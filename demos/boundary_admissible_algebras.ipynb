# $\widehat{\mathfrak{su}}(3)_{-9/4}$

```sage
alg = AffineLieAlgebra(["A", 2, 1])

α = alg.finite_lie_algebra.simple_roots()
αhat = alg.affine_simple_roots()
ω = alg.finite_lie_algebra.fundamental_weights()
ωhat = alg.fundamental_weights()

rho = alg.finite_lie_algebra.rho()
rho_hat = alg.rho_hat()
theta = alg.finite_lie_algebra.theta()
theta_hat = alg.theta_hat()

delta = alg.delta()

What = alg.affine_weyl_group()

```

```sage
# Find all the admissible highest weight at level k = -3 + 3/4
kF = -3 + 3/4
uu = kF.denominator()
tt = kF.numerator()

t = alg.extended_affine_weyl_group().translation
s = alg.extended_affine_weyl_group().simple_reflection

Pi_u = [uu * delta - theta_hat] + [alg.affine_simple_roots()[i] for i in range(1,alg.rank)]
Pi_u
```

```sage
# list the Weyl group elements as semi-direct product of the finite Weyl group and the coweight lattice
W_elements = alg.extended_affine_weyl_group().elements_as_semi_direct_product(
  translation_bounds={1: (-5, 6), 2: (-5, 6)},
  factor_order="ts"
)
W_elements
```

```sage
# t_{-ω1}
[is_positive_affine_root(t(-alg.fundamental_coweights()[1]).action(weight)) for weight in Pi_u]

# t_{-2ω1}
[is_positive_affine_root(
    t(-2*alg.fundamental_coweights()[1]).action(weight)) for weight in Pi_u]

# t_{-2ω1}
[is_positive_affine_root(
    t(-2*alg.fundamental_coweights()[1]).action(weight)) for weight in Pi_u]

# t_{- 2ω1 - ω2}
[is_positive_affine_root(
    t(-2*alg.fundamental_coweights()[1] - alg.fundamental_coweights()[2]).action(weight)) for weight in Pi_u]


# t_{- 3ω1}
[is_positive_affine_root(
    t(-3*alg.fundamental_coweights()[1]).action(weight)) for weight in Pi_u]

# t_{-ω1 - ω2}
[is_positive_affine_root(
    t(-alg.fundamental_coweights()[1] - alg.fundamental_coweights()[2]).action(weight)) for weight in Pi_u]

# t_{-ω1 - 2ω2}
[is_positive_affine_root(
    t(-alg.fundamental_coweights()[1] - 2*alg.fundamental_coweights()[2]).action(weight)) for weight in Pi_u]
```

# $W_{- 9/4}(\mathfrak{su}(3), [2,1])$

```sage
# Some basic elements about the nilpotent orbit
# l 指代 Levi subalgebra
# f = E^θ
orbit = alg.finite_lie_algebra.nilpotent_orbits()[1]
(x,e,f) = [orbit.sl2_triple().x, orbit.sl2_triple().e,
           orbit.sl2_triple().f]
h = orbit.sl2_triple().h
[x,e,f,h]
# h_f = {h | [h, f] = 0}
# 输出 αv_1 - αv_2，说明对应 h^1 - h^2 in Chevalley basis
h_f = orbit.sl2_triple().h_f()

# 构造 Δ_l = {α ∈ Δ | (α, h) = 0 for all h such that [h, f] = 0}
Delta_l = orbit.sl2_triple().delta_l()
levi = orbit.levi_subalgebra()    # nilpotent orbit 对应的 Levi subalgebra
                                  # 定义为 Delta_l 作为根系对应的李子代数

# 检查 Δ_l 与 h_f 的正交关系
[[alg.scalar_product(h, α) for h in h_f] for α in Delta_l]

# 上述 Δ_l 包含 α1 + α2, -(α1+α2)，都不是原本李代数所选的标准单根，因此相对于标准单根，这个
# Levi subalgebra l 不是 standard Levi
# 换言之，f 不是 standard Levi 代数的 regular element

# 可以通过 conjugate 将 levi 转到 standard Levi
# 或者通过 conjugate 将 (x,e,f) -> (x', e', f')，使得 f' 是 standard Levi 代数的 regular element

# check the relation between f = E^θ and two different Levi subalgebras

orbit.is_regular(levi)  # E^θ 是 θ,-θ Δ_l 对应的 Levi 代数的 regular element
orbit.is_regular([-α[1], α[1]])  # E^θ 不是 a1,-α1 对应的 Levi 代数的 regular element
```

```sage
# find all t_βy such that t_βy(Pi_u) is positive affine but in Δ_l
tβys = [w for w in W_elements if all(
    [is_positive_affine_root(w.action(weight)) and not is_zero_weight(w.action(weight) - alg.affine_simple_roots()[1]) for weight in Pi_u])]

# the admissible higheest weights of the quantum reduction is given by t_βy(kF Λ_0)
admissible_highest_weights = list(dict.fromkeys(
    [w.action(kF * alg.fundamental_weights()[0] + rho_hat) - rho_hat for w in tβys]))
# there are 12 admssible highest weights
print(len(admissible_highest_weights))

# 12 个 weights 里面只有 6 个是实际不同的，因为 Λ' in W_f.Λ 的两个 weight Λ', Λ 给出相同的 reduction

```
