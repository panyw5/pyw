# Nilpotent Orbits of Classical Lie Algebras

This notebook demonstrates the `nilpotent_orbits` module, which enumerates nilpotent orbits
for classical Lie algebras of type A, B, C, D via the **partition correspondence**.

## Background

Let $\mathfrak{g}$ be a complex semisimple Lie algebra and $G$ its adjoint group.
An element $e \in \mathfrak{g}$ is **nilpotent** if $\mathrm{ad}(e)$ is a nilpotent endomorphism.
The set of all nilpotent elements forms the **nilpotent cone** $\mathcal{N} \subset \mathfrak{g}$,
which decomposes into finitely many $G$-orbits.

For classical types, these orbits are classified by **partitions** (Collingwood & McGovern, 1993):

| Type | Lie algebra | Partition of | Condition |
|------|-------------|-------------|----------|
| $A_n$ | $\mathfrak{sl}(n+1)$ | $n+1$ | None |
| $B_n$ | $\mathfrak{so}(2n+1)$ | $2n+1$ | Even parts have even multiplicity |
| $C_n$ | $\mathfrak{sp}(2n)$ | $2n$ | Odd parts have even multiplicity |
| $D_n$ | $\mathfrak{so}(2n)$ | $2n$ | Even parts have even multiplicity; very even $\to$ 2 orbits |

## 1. Type A: $\mathfrak{sl}(N)$

Nilpotent orbits of $\mathfrak{sl}(N)$ are in bijection with **partitions of $N$**.
Each partition $\lambda = [\lambda_1, \lambda_2, \ldots]$ corresponds to a nilpotent matrix
with Jordan blocks of sizes $\lambda_1, \lambda_2, \ldots$

The orbit dimension is
$$
\dim \mathcal{O}_\lambda = N^2 - \sum_i (\lambda_i^t)^2
$$
where $\lambda^t$ is the transpose (conjugate) partition.

````sage
# sl(5) = A4: partitions of 5 → 7 orbits
print(nilpotent_orbit_table('A', 4))
````

````sage
alg = AffineLieAlgebra(['A', 2, 1])
````

````sage
# Access individual orbit data
orbits_A4 = nilpotent_orbits('A', 4)

print(f"Number of orbits: {len(orbits_A4)}")
print()

# Zero orbit (trivial): partition [1,1,1,1,1]
zero = orbits_A4[0]
print(f"Zero orbit:    {zero.partition}, dim = {zero.dimension}")

# Regular orbit (dense in nilpotent cone): partition [5]
regular = orbits_A4[-1]
print(f"Regular orbit: {regular.partition}, dim = {regular.dimension}")
print(f"  dim = N^2 - N = 25 - 5 = {5**2 - 5}")
````

### Transpose partition and duality

In type A, the **Barbasch-Vogan duality** (Spaltenstein duality) is simply
the transpose of the partition: $d(\mathcal{O}_\lambda) = \mathcal{O}_{\lambda^t}$.

````sage
for orb in nilpotent_orbits('A', 4):
    print(f"  {str(orb.partition):<20}  <-->  {str(orb.transpose):<20}  (dim {orb.dimension})")
````

## 2. Type B: $\mathfrak{so}(2n+1)$

Nilpotent orbits of $\mathfrak{so}(2n+1)$ correspond to partitions of $2n+1$
where **even parts occur with even multiplicity**.

````sage
# so(7) = B3
print(nilpotent_orbit_table('B', 3))
````

## 3. Type C: $\mathfrak{sp}(2n)$

Nilpotent orbits of $\mathfrak{sp}(2n)$ correspond to partitions of $2n$
where **odd parts occur with even multiplicity**.

````sage
# sp(6) = C3
print(nilpotent_orbit_table('C', 3))
````

## 4. Type D: $\mathfrak{so}(2n)$

Nilpotent orbits of $\mathfrak{so}(2n)$ correspond to partitions of $2n$
where **even parts occur with even multiplicity**, with one subtlety:

> A **very even partition** (all parts are even) gives rise to **two** distinct orbits,
> labelled I and II.

This reflects the fact that $SO(2n)$ has two connected components in the stabilizer.

````sage
# so(8) = D4 (triality!)
print(nilpotent_orbit_table('D', 4))
````

````sage
# Inspect the very-even orbits in D4
orbits_D4 = nilpotent_orbits('D', 4)
very_even = [o for o in orbits_D4 if o.is_very_even]

print(f"Very even orbits in D4: {len(very_even)}")
for o in very_even:
    print(f"  {o.label:<20}  dim = {o.dimension}")
````

## 5. Orbit Counts Across Ranks

Let's see how the number of nilpotent orbits grows with rank.

````sage
print(f"{'Type':<8} {'n=1':>5} {'n=2':>5} {'n=3':>5} {'n=4':>5} {'n=5':>5} {'n=6':>5}")
print("-" * 44)
for letter in ['A', 'B', 'C']:
    counts = [len(nilpotent_orbits(letter, n)) for n in range(1, 7)]
    row = f"{letter+'_n':<8}" + "".join(f"{c:>5}" for c in counts)
    print(row)

# D_n starts at n=2
counts_D = ['-'] + [str(len(nilpotent_orbits('D', n))) for n in range(2, 7)]
row = f"{'D_n':<8}" + "".join(f"{c:>5}" for c in counts_D)
print(row)
````

## 6. Orbit Properties: Distinguished and Even

Two important classes of orbits:

- **Distinguished**: the centralizer $Z_G(e)$ is unipotent (all parts distinct).
- **Even**: the weighted Dynkin diagram has only even entries.

The **regular orbit** is always both distinguished and even.
The **zero orbit** is even but not distinguished (except in rank 1).

````sage
# Distinguished orbits of sl(6) = A5
orbits_A5 = nilpotent_orbits('A', 5)

print("Distinguished orbits of A5 (sl(6)):")
for o in orbits_A5:
    if o.is_distinguished:
        print(f"  {str(o.partition):<20}  dim = {o.dimension}")

print()
print("Even orbits of A5 (sl(6)):")
for o in orbits_A5:
    if o.is_even:
        print(f"  {str(o.partition):<20}  dim = {o.dimension}")
````

## 7. API Reference

### `nilpotent_orbits(lie_type, rank)`

Returns a `List[NilpotentOrbit]` sorted by ascending dimension.

```python
# String + rank
orbits = nilpotent_orbits('A', 4)

# Tuple input
orbits = nilpotent_orbits(('B', 3))
```

### `NilpotentOrbit` attributes

| Attribute | Type | Description |
|-----------|------|-------------|
| `partition` | `List[int]` | Partition labelling the orbit |
| `dimension` | `int` | Complex dimension of the orbit |
| `transpose` | `List[int]` | Conjugate partition |
| `is_even` | `bool` | Even orbit flag |
| `is_distinguished` | `bool` | Distinguished orbit flag |
| `is_very_even` | `bool` | (Type D) All parts even |
| `roman_numeral` | `str or None` | 'I' or 'II' for very even in type D |
| `label` | `str` | Human-readable label |

### `nilpotent_orbit_table(lie_type, rank)`

Returns a formatted string table of all orbits.

````sage
# Example: inspecting a single orbit object
orb = nilpotent_orbits('D', 4)[6]
print(f"Orbit:           {orb}")
print(f"  partition:     {orb.partition}")
print(f"  dimension:     {orb.dimension}")
print(f"  transpose:     {orb.transpose}")
print(f"  is_even:       {orb.is_even}")
print(f"  is_distinguished: {orb.is_distinguished}")
print(f"  is_very_even:  {orb.is_very_even}")
print(f"  roman_numeral: {orb.roman_numeral}")
print(f"  label:         {orb.label}")
````
