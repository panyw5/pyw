# $\widehat{D}_{6, -4}$

```python
# initialization
cartanType = ["D", 6, 1]
# QLoad=True will instruct the alg to find and load
# saved file with info on the Kazhdan-Lusztig inverse polynomial
# QLoad=False will skip the loading
# and compute the polynomial on the fly
alg = Alg.Alg(cartanType, QLoad=True)
```

```python
import time
def GetLambda(alg, llambda):
  # lambda is a reserved word, use llambda instead
  # =====================================================================
  # When k + hcheck > 0
  # find the unique weight Lambda of the form "Λ = w(λ + ρ) - ρ"
  # such that Λ + ρ is dominant
  # i.e., Λ + ρ has NON-NEGATIVE (can be zero) Dynkin labels
  # =====================================================================
  # This function relies on alg.W
  # Use alg.GetWeylGroup() to generate a reasonable alg.W
  start = time.time()
  print(">>> Creating Lambda.", flush=True)
  # If λ itself is already dominant
  # then λ+ρ = id(w + ρ) must be dominant
  # Hence Λ = λ
  alg.llambda = llambda
  if all([i > 0 for i in alg.RemoveDelta(llambda).coefficients()]):
    alg.wToLambda = alg.one
    alg.wTollambda = alg.one
    print("wTollambda = ", alg.wTollambda,
          ", Λ = ", alg.llambda, flush=True)
    return llambda

  # for general cases
  print("Finding Λ for generic λ", flush=True)
  W = alg.WW
  rho = alg.rho
  orbitUnderWeylDot = [(w.action(llambda + rho) - rho) for w in W]
  for weight in orbitUnderWeylDot:
    # .coefficients only collect non-zero Dynkin labels
    reducedCoefficients = alg.RemoveDelta(weight).coefficients()
    if all(c >= -1 for c in reducedCoefficients):
      alg.Lambda = weight
      index = orbitUnderWeylDot.index(weight)
      alg.wToLambda = W[index]         # wToLamda(λ + ρ) - ρ = Λ
                                    # wToLambda(λ+ρ) = Λ+ρ
      alg.wTollambda = alg.wToLambda.inverse()
      print("wTollambda = ", alg.wTollambda,
            ", Λ = ", alg.Lambda, flush=True)
                                    # λ+ρ = wToLambda^{-1}(Λ+ρ)
      end = time.time()
      print(">>> Lambda created: %s s" %
            str(end - start), "\n", flush=True)
      return weight
  raise Exception("No Lambda found: increase the size of Weyl group")
```
